# Residence Hall Maintenance Request Analysis

This notebook is a lightweight walkthrough of the reproducible pipeline in `src/`. All data is synthetic and fictionalized.

```python
from pathlib import Path
import json
import pandas as pd

ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
clean_path = ROOT / 'data' / 'processed' / 'maintenance_requests_clean.csv'
df = pd.read_csv(clean_path, parse_dates=['submitted_at', 'assigned_at', 'first_response_at', 'resolved_at'])
df.head()
```

```python
metrics = json.loads((ROOT / 'reports' / 'metrics.json').read_text())
pd.Series(metrics, name='value')
```

```python
category_summary = (
    df.groupby('category', as_index=False)
      .agg(requests=('request_id', 'count'), avg_assignment_hours=('assignment_hours', 'mean'))
      .sort_values('requests', ascending=False)
)
category_summary
```
